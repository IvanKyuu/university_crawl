```python
import threading
from concurrent.futures import ThreadPoolExecutor

result_dict = {}


attr_to_run = [
    "university_type",
    "graduation_year",
    "location",
    "graduation_rate",
    "domestic_student_tuition",
    "international_student_tuition",
    "program_official_website",
    "program_enrollment_number",
]
```

```python
import threading 

for attr in attr_to_run:
    # Ensure the column is of type 'object' (string)
    if attr in merged_data.columns:
        merged_data[attr] = merged_data[attr].astype('object')
    else:
        # If the column doesn't exist, create it with type 'object'
        merged_data[attr] = ""
```

```python
# Function to process attributes with threading
def process_attribute(attr, value, uni_name, program_name, faculty_name, result_dict):
    attr_format = value["attribute_format"]
    attr_prompt = value["attribute_prompt"]
    example = value["example"]
    reference = value["attribute_reference"]
    result = uni_program_gen.get_program_info(
        university_name=uni_name,
        program_name=program_name,
        faculty_name=faculty_name,
        target_attribute=attr,
        format_=attr_format,
        reference=reference,
        _data_example_pair=example,
        _extra_prompt=attr_prompt,
    )
    result_dict[attr] = result

# Iterate over each row in the DataFrame
count = 0
for idx in merged_data.index:
    uni_name = merged_data.loc[idx, "university_name"]
    faculty_name = merged_data.loc[idx, "faculty_name"]
    program_name = merged_data.loc[idx, "program_name"]
    program_id_ = merged_data.loc[idx, "id_"]

    threads = []
    result_dict = {}

    for attr, value in attr_dict.items():
        if pd.isna(value["handler"]) or not value["handler"]:
            continue
        elif attr in attr_to_run:
            # print("* " * 20 + "attr " + attr)
            # Create a thread for each attribute
            thread = threading.Thread(target=process_attribute, args=(attr, value, uni_name, program_name, faculty_name, result_dict))
            threads.append(thread)
            thread.start()

    # Join all threads to ensure completion
    for thread in threads:
        thread.join()

    # Update the DataFrame with the results
    for attr in result_dict:
        if attr == "program_enrollment_number":
            print("* " * 20 + "program_enrollment_number " + result_dict[attr])
        merged_data.loc[idx, attr] = result_dict[attr]
    
    if count >= 1000:
        break
    count += 1
```
