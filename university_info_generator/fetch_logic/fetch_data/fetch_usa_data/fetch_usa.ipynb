```python
# usa_universities
client = OpenAI(api_key="")

prompt = """
## Instruction
check if a university is an ivy league, if it is, then return 3 famous people that graduated from the university:
one politician
one CEO level famous people
one celebrity

Example:
Yale University
George H.W. Bush: 41st U.S. President
Meryl Streep: Acclaimed actress
Hillary Clinton: Former Secretary of State
# """

messages = [
            {"role": "system", "content": prompt},
            {
                "role": "user",
                "content": f"university_name: Cornell University",
            },
        ]
```

```python
response = client.chat.completions.create(messages=messages, model="gpt-4o", temperature=0.0)
result = response.choices[0].message.content
```
