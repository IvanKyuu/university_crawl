```python
df = pd.read_csv("/home/ivan/Uforse/university_crawl/university_info_generator/fetch_logic/fetch_data/all_universities_usa_info.csv")
```

```python
count = 0
for ind in tqdm(df.index):
    # if count >= 10:
    #     break
    count += 1
    # print(df["university_name"][ind], df["ranking_us_news_2023"][ind])
    uni_name = df["university_name"][ind]
    r = requests.get(url=url+"+".join(uni_name.split()), headers=headers)
    if r.status_code != 200:
        break
    soup = BeautifulSoup(r.content, "html.parser")
    li = soup.find(name="ol").find_next(name="li")
    if li:
        rank = li.find_next(name="strong")
    else:
        rank = None
    if rank:
        rank = rank.text
        if rank[0] == "#":
            rank = rank[1:]
    else:
        rank = ""
    df.at[ind, "ranking_us_news_2023"] = rank
    # print(df["university_name"][ind], df["ranking_us_news_2023"][ind])
```
