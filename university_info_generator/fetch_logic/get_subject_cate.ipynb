```python
from selenium import webdriver
import requests 
```

```python
url = "https://www.topuniversities.com/subject-rankings"
page = requests.get(url)
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(page.content, "html.parser")
soup.prettify()
```

```python
lst = soup.find(name="div", attrs={"class": "layout__region layout__region--content"}).findNext(name="div", attrs={"class": "layout__region layout__region--content"}).find_all(
    name="div", attrs={"class": "content"}, recursive=True
)
```

```python
print(len(lst))
```

```python
result_dict = {}
```

```python
result_dict = {}
for general_subject_card in lst:
    if not general_subject_card:
        continue
    general_subject_name = general_subject_card.find("h2", recursive=True)
    if not general_subject_name:
        continue
    general_subject_name = general_subject_name.text.lower()
    tab = general_subject_card.find(name="div", attrs={"class": "select-subj"})
    if not tab:
        continue
    subject_lst = tab.find_all(name="li")
    # print(general_subject_name)
    # print(subject_lst)
    result_dict[general_subject_name] = set()
    for subject in subject_lst:
        text = subject.text
        text = text.strip().strip("(2024)").lower()
        result_dict[general_subject_name].add(text)
```

```python
from pprint import pprint
pprint(result_dict)
```

```python
import pandas as pd
data_tuples = []
for category, subjects in result_dict.items():
    for subject in subjects:
        data_tuples.append((category, subject))

# Create a DataFrame from the list of tuples
df = pd.DataFrame(data_tuples, columns=['Category', 'Subject'])



# import ace_tools as tools; tools.display_dataframe_to_user(name="University Subjects", dataframe=df)

df.head()
```
